```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver import Chrome
import time , csv
urls = [
    "https://www.ttt.live/ministry-of-the-people-distributes-over-69000-in-seed-grants-to-san-fernando-entrepreneurs/",
    "https://www.ttt.live/10-held-in-foiled-plot-to-harm-law-enforcement-officer/",
    "https://www.ttt.live/ttuta-police-in-schools-must-be-temporary/",
    "https://www.ttt.live/major-law-enforcement-operation-summer-slam-nets-32-arrests-firearms-narcotics-and-ammunition/",
    "https://www.ttt.live/anti-crime-exercise-yields-seizure-of-high-powered-weapon-in-diego-martin/",
    "https://www.ttt.live/seven-arrested-during-soe-operation-in-point-fortin/",
    "https://www.ttt.live/ministers-back-us-military-efforts-to-curb-regional-drug-and-arms-trade/",
    "https://www.ttt.live/man-charged-with-murder-in-connection-to-death-of-ravi-pandhorie/",
    "https://www.ttt.live/five-held-two-firearms-ammunition-and-narcotics-seized-in-soe-operations/",
    "https://www.ttt.live/police-commissioner-supports-us-strike-issues-warning-to-fishers-of-fortune/"
    ]

options= webdriver.ChromeOptions()
options.add_argument("--headers")
options.page_load_strategy = "none"
driver = webdriver.Chrome(options=options)
driver.implicitly_wait(5)
with open("tttnews_data.csv","w",newline="",encoding="utf-8") as f:
    writer = csv.writer(f)
    writer.writerow(["Title","Article","Author","Date"])
   
    for url in urls:
        driver.get(url)
        time.sleep(10)
        article=driver.find_element(By.CLASS_NAME,"entry-content")
        title=driver.find_element(By.CLASS_NAME,"entry-title")
        date=driver.find_element(By.CLASS_NAME,"date")
        author=driver.find_element(By.CLASS_NAME,"by-author")
        writer.writerow([title.text,article.text,author.text,date.text])
        
    
driver.close()
```
